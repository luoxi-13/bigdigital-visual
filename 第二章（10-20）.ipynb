```python
import matplotlib.pyplot as plt
import numpy as np

# 1. 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False

# 2. 创建画布
fig, ax = plt.subplots(figsize=(12, 7))
bg_color = '#151932'  # 深蓝色背景
fig.patch.set_facecolor(bg_color)
ax.set_facecolor(bg_color)

# 3. 绘制核心图形 (水位圆环)
R = 0.35  # 圆环半径
fill_ratio = 0.65  # 完成度 65%

# 计算水位线高度 (圆心为0,0，底部为-R，顶部为R)
# y_level = -R + 2*R*fill_ratio
y_level = -R + 2 * R * fill_ratio

# 计算水位线与圆的交点角度
# sin(alpha) = y_level / R
alpha_rad = np.arcsin(y_level / R)
alpha_deg = np.degrees(alpha_rad)

# 左侧交点角度 (第二象限) 和 右侧交点角度 (第一象限，转换为 360+ 以便顺时针/逆时针连续)
start_angle = 180 - alpha_deg  # 约 162.5 度
end_angle = 360 + alpha_deg    # 约 377.5 度

# 生成下半部分圆弧的点 (从左边交点，经过底部，到右边交点)
theta = np.linspace(np.radians(start_angle), np.radians(end_angle), 200)
x_circle = R * np.cos(theta)
y_circle = R * np.sin(theta)

# 构造多边形顶点 (首尾相连，顶部自动形成水平直线)
poly_x = np.concatenate([x_circle, [x_circle[0]]])
poly_y = np.concatenate([y_circle, [y_circle[0]]])

# 填充颜色 (亮蓝色，接近原图)
fill_color = '#0072c6' 
ax.fill(poly_x, poly_y, color=fill_color, zorder=5)

# 绘制外圆环轮廓 (细细的蓝色线)
circle_outline = plt.Circle((0, 0), R, fill=False, edgecolor='#0072c6', linewidth=2, zorder=6)
ax.add_patch(circle_outline)

# 4. 添加中心文字 "65%"
# 位置稍微偏下，位于蓝色区域内
ax.text(0, -0.05, '65%', color='white', fontsize=60, fontweight='bold', 
        ha='center', va='center', fontfamily='Arial', zorder=10)

# 5. 标题 (左上角，加粗加大)
fig.text(0.1, 0.90, '2022年上半年目标完成率', color='white', fontsize=32, fontweight='bold', ha='left')
fig.text(0.1, 0.81, '截至6月30日销售目标总体完成率达到65%', color='white', fontsize=18, ha='left')

# 6. 底部注释
fig.text(0.1, 0.08, '*注：数据来源于公司销售系统', color='#aaaaaa', fontsize=12, ha='left')

# 7. 调整布局与坐标轴
ax.set_xlim(-1, 1)
ax.set_ylim(-0.5, 0.5)
ax.set_aspect('equal')  # 强制正圆
ax.axis('off')          # 隐藏坐标轴

# 垂直居中布局
plt.subplots_adjust(top=0.85, bottom=0.15, left=0.1, right=0.9)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as patches

# 1. 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False

# 2. 创建画布
fig, ax = plt.subplots(figsize=(12, 7))
bg_color = '#151932'  # 深蓝色背景
fig.patch.set_facecolor(bg_color)
ax.set_facecolor(bg_color)

# 3. 参数设置
R = 0.35  # 圆环半径
fill_ratio = 0.65  # 占比 65%
fill_color = '#0072c6'  # 亮蓝色
ring_color = '#0072c6'  # 圆环颜色（同色或稍浅）

# 4. 绘制外圆环 (细细的蓝色线)
# 这是一个空心圆，作为容器边界
circle_ring = plt.Circle((0, 0), R, fill=False, edgecolor=ring_color, linewidth=2, zorder=5)
ax.add_patch(circle_ring)

# 5. 绘制波浪水位 (核心难点)
# 计算水位基准线高度
# 圆底部是 -R，顶部是 R，总高度 2R。
# 65% 高度 = -R + 2R * 0.65 = 0.3R
base_level = -R + 2 * R * fill_ratio 

# 生成波浪数据
# x 范围要比圆宽一点 (-1.2R 到 1.2R)，防止裁剪边缘露馅
x_wave = np.linspace(-1.2 * R, 1.2 * R, 500)
# 波浪公式：y = 基准线 + 振幅 * sin(频率 * x)
# 振幅设为 0.05 (比较平缓)，频率设为 10 (大概两个波峰)
amplitude = 0.04 
frequency = 8 
y_wave = base_level + amplitude * np.sin(frequency * x_wave)

# 构造封闭多边形用于填充 (底部封死)
# 顶点顺序：从左到右沿波浪走 -> 右下角 -> 左下角 -> 回到起点
poly_x = np.concatenate([x_wave, [1.2 * R, -1.2 * R]])
poly_y = np.concatenate([y_wave, [-1.5 * R, -1.5 * R]]) # 底部延伸到圆外

# 创建圆形裁剪路径 (Clip Path)
# 这样波浪只会显示在圆里面
clip_circle = plt.Circle((0, 0), R - 0.01, transform=ax.transData) # 稍微小一点点，避免压住边框

# 绘制填充
# 使用 zorder=4 让它在圆环下面，或者 zorder=6 在上面？
# 原图看起来蓝色填充在圆环内部。
wave_patch = ax.fill(poly_x, poly_y, color=fill_color, zorder=4)
# 应用裁剪
for patch in wave_patch:
    patch.set_clip_path(clip_circle)

# 6. 添加中心文字 "65%"
# 位置在圆心附近
ax.text(0.02, 0.02, '65%', color='white', fontsize=60, fontweight='bold', 
        ha='center', va='center', fontfamily='Arial', zorder=10)

# 7. 标题 (左上角，加粗加大)
fig.text(0.1, 0.90, '本科及以上学历员占比', color='white', fontsize=32, fontweight='bold', ha='left')
fig.text(0.1, 0.81, '6月30日最新统计数据本科及以上员工占比65%', color='white', fontsize=18, ha='left')

# 8. 底部注释
fig.text(0.1, 0.08, '*注：数据来源于公司人力资源系统', color='#aaaaaa', fontsize=12, ha='left')

# 9. 调整布局与坐标轴
ax.set_xlim(-1, 1)
ax.set_ylim(-0.5, 0.5)
ax.set_aspect('equal')  # 强制正圆
ax.axis('off')          # 隐藏坐标轴

# 垂直居中布局
plt.subplots_adjust(top=0.85, bottom=0.15, left=0.1, right=0.9)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
# 1. 字体设置
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False
# 2. 创建画布，采用 4:3 左右的比例
fig, ax = plt.subplots(figsize=(10, 8), facecolor='#151A35') 
ax.set_facecolor('#151A35')
ax.axis('off')
# 锁定坐标轴范围，确保饼图是正圆，且留出足够的空间画引导线
ax.set_xlim(-2.2, 2.2)
ax.set_ylim(-1.6, 1.6)
ax.set_aspect('equal')
# 3. 绘制饼图
# 数据：销售部29.2%，采购部25.0%，工程部18.0%，财务部14.0%，行政部8.0%，人力部5.8%
sizes = [29.2, 25.0, 18.0, 14.0, 8.0, 5.8]
labels = ["销售部", "采购部", "工程部", "财务部", "行政部", "人力部"]
colors = ['#FF4F64', '#F5C355', '#22B4AE', '#2E7DB9', '#2F9ED0', '#653FA3']
# 起始角度设为90度(12点方向)，逆时针设为False(顺时针旋转)
wedges, _ = ax.pie(sizes, colors=colors, startangle=90, counterclock=False, 
                   radius=0.8, wedgeprops=dict(edgecolor='#151A35', linewidth=1.5))

# 4. 绘制特色引导线 (折线 + 起点小圆点)
def draw_guide_line(wedge, color, points, text_label):
    # 画线
    xs = [p[0] for p in points]
    ys = [p[1] for p in points]
    ax.plot(xs, ys, color=color, linewidth=1.5, zorder=2)
    # 画起点小圆点（饼图扇区上）
    ax.scatter(xs[0], ys[0], color=color, s=40, zorder=3, edgecolors='white', linewidth=0.5)
    # 文字放在引导线末端
    ax.text(xs[-1]+0.05, ys[-1], text_label, color='white', fontsize=12, va='center')

# --------------------------6个部门引导线（完全对齐左侧原图样式）--------------------------
# 红色 销售部(29.2%) - 右侧折线
draw_guide_line(wedges[0], colors[0], [(0.8, 0.25), (1.15, 0.25), (1.25, 0.5), (1.6, 0.5)], "销售部 29.2%")
# 黄色 采购部(25%) - 右下折线
draw_guide_line(wedges[1], colors[1], [(0.75, -0.82), (0.9, -0.92), (1.4, -0.92)], "采购部 25.0%")
# 青色 工程部(18%) - 左下折线
draw_guide_line(wedges[2], colors[2], [(-0.48, -0.58), (-0.72, -0.88), (-0.82, -0.72), (-1.5, -0.72)], "工程部 18.0%")
# 深蓝 财务部(14%) - 左侧水平直线
draw_guide_line(wedges[3], colors[3], [(-0.8, -0.22), (-1.5, -0.22)], "财务部 14.0%")
# 浅蓝 行政部(8%) - 左上水平直线
draw_guide_line(wedges[4], colors[4], [(-0.7, 0.52), (-1.5, 0.52)], "行政部 8.0%")
# 紫色 人力部(5.8%) 
draw_guide_line(wedges[5], colors[5], [(-0.3, 0.78), (-1.5, 0.78)], "人力部 5.8%")

# 5. 添加标题和副标题
ax.text(-2.1, 1.4, '2021年各部门人数分布', 
        color='white', fontsize=38, fontweight='black', ha='left')
ax.text(-2.1, 1.25, '公司总人数1664，销售部人数最多451，占比29.2%', 
        color='white', fontsize=18, fontweight='bold', ha='left')

# 6. 添加底部备注
ax.text(-2.1, -1.45, '*注：数据来源于公司人力资源系统，统计日期截至2022.01.01', 
        color='#8E92A8', fontsize=11, ha='left')

# 调整边距
plt.subplots_adjust(left=0.02, right=0.98, top=0.98, bottom=0.02)
# 如果需保存，取消下面注释
# plt.savefig('pie_chart.png', dpi=300, facecolor='#151A35', bbox_inches='tight')
plt.show()
```

```python
# -*- coding: utf-8 -*-
"""1:1 复现「2022年上半各年龄段人数分布」深蓝底环形图（画布 1100×824px，1数据单位=1原图像素）"""
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.patches import Wedge, Arc, Circle
 
# ---------- 中文字体 ----------
matplotlib.rcParams['font.family'] = 'sans-serif'
matplotlib.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei',
                                          'PingFang SC', 'Noto Sans CJK SC']
matplotlib.rcParams['axes.unicode_minus'] = False
 
# ---------- 原图配色 ----------
BG, PURPLE, BLUE = '#161C30', '#6C4FE0', '#4B7BF5'
CYAN, ORANGE, GRAY = '#52C9D9', '#F2B14E', '#8F97A8'
LIGHT = '#E9ECF5'
 
# ---------- 画布 ----------
fig = plt.figure(figsize=(11, 8.24), dpi=100)
fig.patch.set_facecolor(BG)
ax = fig.add_axes([0, 0, 1, 1])
ax.set_facecolor(BG); ax.axis('off')
ax.set_xlim(0, 1100); ax.set_ylim(0, 824); ax.set_aspect('equal')
 
# ---------- 环形图 ----------
cx, cy, r_in = 581, 342, 140          # 圆心与内径
segs = [                              # (颜色, 外径, 起角, 终角)；90°=正上方，顺时针
    (PURPLE, 216, -45,     90),       # [20,30) 37.5%
    (BLUE,   212, -150.12, -45),      # [30,40) 29.2%
    (CYAN,   208, -225,  -150.12),    # [40,50) 20.8%
    (ORANGE, 178, -270,   -225),      # >=50    12.5%
]
 
def blend(c, t):                      # 向背景色混合，用于同心细纹
    a, b = matplotlib.colors.to_rgb(c), matplotlib.colors.to_rgb(BG)
    return tuple(a[i]*(1-t) + b[i]*t for i in range(3))
 
for color, r_out, t1, t2 in segs:
    ax.add_patch(Wedge((cx, cy), r_out, t1, t2, width=r_out-r_in,
                       facecolor=color, edgecolor=BG, linewidth=2.5))
    for f in (0.28, 0.52, 0.76):      # 环内 3 条同心细弧
        r = r_in + (r_out - r_in) * f
        ax.add_patch(Arc((cx, cy), 2*r, 2*r, theta1=t1+1, theta2=t2-1,
                         lw=1.3, color=blend(color, 0.38), zorder=2))
 
# ---------- 折线引导标注 ----------
def annotate(color, cat, pct, dot, elbow, x0, x1, yline, tx, ty, pct_color=None):
    ax.plot([elbow[0], dot[0]], [elbow[1], dot[1]], color=GRAY, lw=2,
            solid_capstyle='round', zorder=5)                    # 灰色斜线
    ax.plot([x0, x1], [yline, yline], color=color, lw=2.2,
            solid_capstyle='round', zorder=5)                    # 彩色横线
    ax.add_patch(Circle(dot, 5.5, facecolor=color, edgecolor='none', zorder=6))
    ax.text(tx, ty, cat, color=LIGHT, fontsize=14, ha='left', va='top')
    ax.text(tx, ty-29, pct, color=pct_color or color, fontsize=15,
            fontweight='bold', ha='left', va='top')
 
annotate(PURPLE, '[20,30)', '37.5%', (820, 479), (852, 527), 852, 1002, 527, 885, 581)
annotate(BLUE,   '[30,40)', '29.2%', (712, 132), (772,  82), 772,  925,  82, 808, 141)
annotate(CYAN,   '[40,50)', '20.8%', (336, 374), (300, 332), 120,  300, 332, 185, 389)
annotate(ORANGE, '>=50',    '12.5%', (434, 524), (390, 568), 228,  390, 568, 302, 632,
         pct_color=LIGHT)
 
# ---------- 标题 / 副标题 / 脚注 ----------
ax.text(85, 786, '2022年上半各年龄段人数分布',
        color='white', fontsize=29, fontweight='bold', va='top')
ax.text(85, 719, '公司平均年龄32.5， 20-30员工比例最高占比37.5%',
        color='#F2F4FA', fontsize=18, va='top')
ax.text(85, 66, '*注：数据来源于公司人力资源系统，统计日期截至2022.06.30',
        color='#CBD0DC', fontsize=13.5, va='top')
 
plt.savefig('age_donut.png', dpi=100, facecolor=BG)   # 输出 1100×824，与原图同尺寸
plt.show()
```
